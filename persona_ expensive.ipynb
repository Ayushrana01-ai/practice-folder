```python
import matplotlib.pyplot as plt

subjects = ["Python", "SQL", "Statistics", "ML", "AI"]
marks = [85, 78, 90, 75, 88]

total = sum(marks)
percentage = total / len(marks)

print("Total Marks:", total)
print("Percentage:", percentage, "%")

if percentage >= 80:
    print("Grade: A")
elif percentage >= 60:
    print("Grade: B")
else:
    print("Grade: C")

plt.bar(subjects, marks)
plt.title("Student Marks Analysis")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.ylim(0, 100)
plt.show()
```
